```python
from pathlib import Path
import pandas as pd

df = pd.read_csv(r'C:\Users\kalav\OneDrive\Desktop\Car-Price-Predictor\data\raw\CAR DETAILS FROM CAR DEKHO.csv')

duplicate_mask = df.duplicated(keep=False)

print("Total rows:", len(df))
print("Exact duplicate rows beyond first occurrences:",
      df.duplicated().sum())
print("Rows involved in duplicate groups:", duplicate_mask.sum())

display(
    df.loc[duplicate_mask]
      .sort_values(list(df.columns))
      .head(20)
)
```

```python
print("Manufacturing year counts:")
display(df["year"].value_counts().sort_index())

print("\nLowest 10 selling prices:")
display(df.nsmallest(10, "selling_price"))

print("\nHighest 10 selling prices:")
display(df.nlargest(10, "selling_price"))

print("\nHighest 10 kilometre readings:")
display(df.nlargest(10, "km_driven"))
```

```python
categorical_columns = [
    "fuel",
    "seller_type",
    "transmission",
    "owner"
]

for col in categorical_columns:
    print(f"\n{col} — unique values and counts")
    display(df[col].value_counts(dropna=False))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.hist(df["selling_price"], bins=40, edgecolor="black")
plt.title("Distribution of Used-Car Selling Prices")
plt.xlabel("Selling Price")
plt.ylabel("Number of Cars")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```
