```python
import pandas as pd
import numpy as np
from pathlib import Path


df = pd.read_csv(r'C:\Users\kalav\OneDrive\Desktop\Car-Price-Predictor\data\raw\CAR DETAILS FROM CAR DEKHO.csv')

print("Original dataset shape:", df.shape)
display(df.head())
```

```python
print("Missing values:")
display(df.isnull().sum())

print("Exact duplicate rows:", df.duplicated().sum())

print("Data types:")
display(df.dtypes)
```

```python
rows_before = len(df)

df = df.drop_duplicates().copy()

print("Rows before cleaning:", rows_before)
print("Duplicate rows removed:", rows_before - len(df))
print("Rows after cleaning:", len(df))
print("Remaining duplicates:", df.duplicated().sum())
```

```python
text_columns = df.select_dtypes(include=["object"]).columns

for col in text_columns:
    df[col] = df[col].str.strip()

print("Text cleaning completed.")
display(df.head())
```

```python
print("Invalid selling prices:", (df["selling_price"] <= 0).sum())
print("Invalid kilometre readings:", (df["km_driven"] <= 0).sum())
print("Years outside the observed 1992–2020 range:",
      (~df["year"].between(1992, 2020)).sum())

display(df[["year", "selling_price", "km_driven"]].describe())
```

```python
OUTPUT_PATH = Path("../data/processed/car_price_cleaned.csv")
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)

df.to_csv(OUTPUT_PATH, index=False)

print("Cleaned dataset saved successfully!")
print("Output path:", OUTPUT_PATH.resolve())
print("Final shape:", df.shape)
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```
